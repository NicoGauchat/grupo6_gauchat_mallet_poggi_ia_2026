```python
# cargamos los tres archivos originales de la competencia
from pathlib import Path
import pandas as pd

archivos_necesarios = ["solicitudes_train.csv", "solicitudes_test.csv", "comportamiento_bancario.csv"]
faltantes = [nombre for nombre in archivos_necesarios if not Path(nombre).is_file()]
if faltantes:
    raise FileNotFoundError("Faltan archivos oficiales en data/: " + ", ".join(faltantes))

train = pd.read_csv("solicitudes_train.csv")
train["Estado_Final"].value_counts()
```

## 2. Análisis de Datos

En esta sección exploramos las variables del dataset para entender cómo se relacionan con el default de un prestamo. Analizamos el motivo del préstamo, los ingresos y calculamos nuevas métricas iniciales como el ratio entre el monto solicitado y el ingreso mensual.

```python
train["Motivo_Prestamo"].value_counts()
```

```python
# para ver si existe alguna relacion interesante entre el motivo de prestamo y
# su cumplimiento

pd.crosstab(
    train["Motivo_Prestamo"],
    train["Estado_Final"],
    normalize="index"
)

# el resultado muestra que si bien son bastante similares, negocio es el mas elevado
# le siguen otros y vehiculo
```

```python
train.groupby("Estado_Final")["Ingreso_Mensual"].mean()
# para ver si las personas con default tienen una cantidad menor significativa de ingresos
# no se ve mucha diferencia
```

```python
train.groupby("Estado_Final")["Ingreso_Mensual"].describe()

# no varian mucho los ingresos por lo tanto creo que esta no seria una feature
```

```python
train.groupby("Estado_Final")["Monto_Solicitado"].describe()

# podemos ver que hay una clara tendencia a default en los prestamos de monto mas elevado
# no significa que todos sean pero este es una feature clara
```

```python
train["Ratio_Monto_Ingreso"] = (
    train["Monto_Solicitado"] / train["Ingreso_Mensual"]
)
# metrica creada para saber cuantos meses de ingreso representa el monto de prestamo de un cliente

train.groupby("Estado_Final")["Ratio_Monto_Ingreso"].describe()
# para ver los resultadoss segun estado del prestamo

# podemos ver claramente que el ratio de las personas con default es mas elevado que las que no
# en promedio los prestamos con default representaban un mes mas de ingreso de la persona que los prestamos pagados
```

```python
train.groupby("Estado_Final")["Tasa_Interes"].describe()

# para ver si hay alguna relacion entre la tasa de interes y el estado del prestamo
# podemos ver que los prestamos en default tienen aprox 1 punto mas de intereses
```
